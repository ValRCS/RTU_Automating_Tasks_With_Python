# 📘 Python Dictionaries (dict)

This notebook explains Python dictionaries step by step.

## Creating an empty dictionary

```
empty_dict_1 = {}
empty_dict_2 = dict()

print(empty_dict_1)
print(type(empty_dict_1))
```

## Adding key–value pairs

```
student = {}
student["name"] = "Anna"
student["age"] = 21
student["phone"] = "+37112345678"

student
```

## Accessing values

```
print(student["name"])
print(student["age"])
```

## Safe access with get()

```
print(student.get("email"))
print(student.get("email", "Not provided"))
```

## Updating values

```
student["age"] = 22
student
```

## Removing items

```
removed_phone = student.pop("phone")
removed_phone
```

```
student
```

## Checking if key exists

```
print("name" in student)
print("email" in student)
```

## Keys, values, items

```
student = {
    "name": "Anna",
    "age": 22,
    "courses": ["Python", "Databases", "Algorithms"]
}

student.keys()
```

```
student.values()
```

```
student.items()
```

## Looping through dictionary

```
for key, value in student.items():
    print(key, "->", value)
```

## Student dictionary – realistic example

```
student = {
    "name": "Anna Ozola",
    "age": 22,
    "email": "anna.ozola@example.com",
    "phone": "+37112345678",
    "courses": ["Python", "Algorithms", "Databases"],
    "graduated": False
}

student
```

## Nested dictionaries

```
student["address"] = {
    "city": "Riga",
    "street": "Brivibas iela",
    "postal_code": "LV-1010"
}

student
```

```
print(student["address"]["city"])
```

## Copying dictionaries

```
student_copy = student.copy()
student_copy["age"] = 24

student, student_copy
```

## List of student dictionaries

```
students = [
    {"name": "Anna", "age": 22, "courses": ["Python", "Algorithms"]},
    {"name": "Janis", "age": 24, "courses": ["Databases", "Web Development"]},
    {"name": "Elina", "age": 21, "courses": ["Python", "Statistics"]}
]

students
```

## Looping over students

```
for student in students:
    print(student["name"], "-", student["age"])
```

## Filtering students

```
for student in students:
    if "Python" in student["courses"]:
        print(student["name"])
```

## Common dictionary mistakes

```
student = {"name": "Anna", "age": 22}
print(student.get("email", "Not provided"))
```

## Dictionaries vs Lists

```
student_list = ["Anna", 22, ["Python", "Algorithms"]]

student_dict = {
    "name": "Anna",
    "age": 22,
    "courses": ["Python", "Algorithms"]
}

student_list, student_dict
```
